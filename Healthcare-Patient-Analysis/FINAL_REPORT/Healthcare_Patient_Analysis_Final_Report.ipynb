# Healthcare Patient Analysis
## Final Project Report

**Tools:** Python, NumPy, Pandas, Matplotlib, Seaborn, Plotly

This project analyzes healthcare patient data to understand patient demographics, medical conditions, treatment patterns, hospital information, and billing trends using Python-based data analysis and visualization.


## 1. Project Overview

The project uses a healthcare dataset containing patient demographic, admission, medical condition, treatment, test-result, hospital, insurance, and billing information.

The analysis was completed through phases covering data understanding, data quality, data cleaning, exploratory analysis, patient analysis, disease analysis, treatment analysis, medical cost analysis, and data visualization.


## 2. Objectives

- Understand the structure and characteristics of the healthcare dataset.
- Identify missing, duplicate, invalid, and inconsistent data.
- Clean and prepare the data for analysis.
- Analyze patient demographics such as age and gender.
- Identify common medical conditions.
- Analyze medication and treatment patterns.
- Examine medical billing and cost patterns.
- Create visualizations to communicate important findings.
- Develop a complete portfolio-ready data analysis workflow.


## 3. Dataset Overview

The dataset contains **10,000 patient records and 15 columns**.

Main columns include:

- Name
- Age
- Gender
- Blood Type
- Medical Condition
- Date of Admission
- Doctor
- Hospital
- Insurance Provider
- Billing Amount
- Room Number
- Admission Type
- Discharge Date
- Medication
- Test Results


## 4. Data Understanding

The initial analysis examined the dataset structure, columns, data types, patient records, and major healthcare attributes.

The dataset contains numerical variables such as `Age`, `Room Number`, and `Billing Amount`, together with categorical variables such as `Gender`, `Medical Condition`, `Insurance Provider`, `Admission Type`, `Medication`, and `Test Results`.


## 5. Data Quality and Cleaning

The data-quality and cleaning phases examined missing values, duplicate records, data types, and invalid values.

Date fields were handled as date values where required, and the project separated data-quality checks from the cleaning stage before performing analysis.


## 6. Exploratory Data Analysis

Exploratory analysis was used to understand distributions and relationships in the dataset.

The analysis included patient age, gender, medical condition, billing amount, and other healthcare attributes. Grouping and frequency-based analysis were used to identify patterns in the patient population.


## 7. Patient Analysis

The patient analysis examined age groups, gender distribution, and billing patterns.

### Age-group distribution

| Age Group | Patients |
|---|---:|
| 0–18 | 164 |
| 19–35 | 2,490 |
| 36–50 | 2,142 |
| 51–65 | 2,270 |
| 65+ | 2,934 |

The **65+** group was the largest group in the analyzed dataset.


## 8. Disease Analysis

The `Medical Condition` field was analyzed using frequency counts and grouping by gender and age.

The six medical conditions represented in the dataset were:

- Asthma
- Cancer
- Hypertension
- Arthritis
- Obesity
- Diabetes

The analysis identified **Asthma as the most frequent medical condition, with 1,708 records**.

Disease-by-gender analysis was also performed to compare male and female patient counts within each medical condition.


## 9. Treatment Analysis

The treatment analysis focused on the `Medication` field and examined medication distribution across patient records.

Medication-related billing patterns were also examined using grouped billing analysis.


## 10. Medical Cost Analysis

Medical billing was analyzed using the `Billing Amount` column.

**Average Billing Amount: 25,516.81**

With 10,000 records, the corresponding total billing represented in the dataset is approximately **255.17 million**.

Billing was also examined by patient age, gender, medical condition, and medication.


## 11. Data Visualization

The visualization phase presented the analytical results using charts.

Visual analysis included:

- Patient age distribution
- Gender distribution
- Medical condition distribution
- Medical-cost distribution
- Billing amount comparisons
- Medical cost by disease

Matplotlib, Seaborn, and Plotly were used during the visualization work.


## 12. Key Findings

1. The dataset contains 10,000 healthcare patient records across 15 attributes.
2. The 65+ age group contains the largest number of records (2,934).
3. Asthma is the most frequent medical condition, with 1,708 records.
4. The average billing amount is approximately 25,516.81.
5. Billing patterns were analyzed across age, gender, medical condition, and medication.
6. Visualization makes the major demographic, disease, and cost patterns easier to interpret.

## 15. Limitations

- The dataset is intended for analytical and learning purposes and should not be treated as a clinical decision-making dataset.
- The analysis describes patterns in the available data and does not establish medical causation.
- Results depend on the quality and representativeness of the dataset.
- Healthcare billing can depend on factors that are not available in the dataset.


## 16. Conclusion

The Healthcare Patient Analysis project demonstrates a complete data-analysis workflow using Python, NumPy, Pandas, and visualization tools.

The project progresses from understanding and cleaning the data to patient, disease, treatment, and medical-cost analysis. It demonstrates practical skills in data cleaning, grouping, statistical analysis, and visualization.

This project is suitable as a **Data Science portfolio project** and can later be extended with statistical testing, predictive modeling, dashboards, or machine-learning applications.
