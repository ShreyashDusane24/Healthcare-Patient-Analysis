# LOADING DATASET :

```python
patient_data = pd.read_csv(r"C:\Users\dusan\OneDrive\Pictures\Documents\PROJECT\Healthcare-Patient-Analysis_02\data\healthcare_dataset.csv")
df = pd.DataFrame(patient_data)
df.head()
```

# Most common diseases

```python
df['Medical Condition'].value_counts().head(1)
```

# Disease by gender

```python
df.groupby('Medical Condition')['Gender'].value_counts()
```

# Disease by age group

```python
bin = [0, 19, 36, 51, 66, 120]
label = ['0-18', '19-35', '36-50', '51-65', '65+']
df['age_group'] = pd.cut(df['Age'],bins=bin,labels=label,right=False)
```

```python
df.groupby('Medical Condition')['age_group'].value_counts()
```

# Disease frequency

```python
df['Medical Condition'].value_counts(normalize=True)*100
```
