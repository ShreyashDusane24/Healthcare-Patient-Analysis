```python
import numpy as np
import pandas as pd 
```

# LOADING THE DATASET :

```python
patient_data = pd.read_csv(r"C:\Users\dusan\OneDrive\Pictures\Documents\PROJECT\Healthcare-Patient-Analysis_02\data\healthcare_dataset.csv")
df = pd.DataFrame(patient_data)
df.head()
```

```python
bin = [0, 19, 36, 51, 66, 120]
label = ['0-18', '19-35', '36-50', '51-65', '65+']
df['age_group'] = pd.cut(df['Age'],bins=bin,labels=label,right=False)
```

# most common treatment :

```python
df['Medication'].value_counts()
```

# Treatment frequency

```python
df['Medication'].value_counts(normalize=True)*100
```

# Treatment by disease

```python
df.groupby('Medical Condition')['Medication'].value_counts()
```

# Treatment by age group

```python
df.groupby('age_group')['Medication'].value_counts()
```

# Treatment by gender

```python
df.groupby('Gender')['Medication'].value_counts()
```

# Treatment vs medical cost

```python
df.groupby('Medication')['Billing Amount'].mean()
```
