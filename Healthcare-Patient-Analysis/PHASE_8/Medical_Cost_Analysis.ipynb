```python
import numpy as np
import pandas as pd  
```

# Loading dataset : 

```python
patient_data = pd.read_csv(r"C:\Users\dusan\OneDrive\Pictures\Documents\PROJECT\Healthcare-Patient-Analysis_02\data\healthcare_dataset.csv")
df = pd.DataFrame(patient_data)
df.head()
```

# total medical cost 

```python
total_med = np.array(df['Billing Amount']).sum()
print(total_med)
```

# average medical cost 

```python
avg_medical_cost = np.array(df['Billing Amount']).mean()
print(avg_medical_cost)
```

# median medical cost 

```python
median_med_cost = df['Billing Amount'].median()
print(median_med_cost)
```

# min medical cost 

```python
min_med_cost = np.array(df['Billing Amount']).min()
print(min_med_cost)
```

# max medical cost 

```python
max_medical_cost = np.array(df['Billing Amount']).max()
print(max_medical_cost)
```

# Cost by age group

```python
bin = [0, 19, 36, 51, 66, 120]
label = ['0-18', '19-35', '36-50', '51-65', '65+']
df['age_group'] = pd.cut(df['Age'],bins=bin,labels=label,right=False)
```

# sum

```python
df.groupby('age_group')['Billing Amount'].sum()
```
