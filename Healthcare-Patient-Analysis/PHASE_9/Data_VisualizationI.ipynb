```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px
```

# LOADING DATASET :

```python
patient_data = pd.read_csv(r"C:\Users\dusan\OneDrive\Pictures\Documents\PROJECT\Healthcare-Patient-Analysis_02\data\healthcare_dataset.csv")
df = pd.DataFrame(patient_data)
df.head()
```

# BAR CHARTs : 

# 1.Patients by disease

```python
bar = df['Medical Condition'].value_counts()
sns.barplot(bar)
plt.title('Patients by disease')
plt.grid(axis='y',linestyle = '--')
plt.xticks(rotation = 45,ha = 'right')
plt.show()
```

# 2.Patients by gender

```python
gender_plot = df['Gender'].value_counts()
sns.barplot(gender_plot)
plt.title('Patients by gender')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# 2.Patients by treatment

```python
treatment = df['Medication'].value_counts()
sns.barplot(treatment)
plt.title('Patients by treatment')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# 3.Average cost by disease

```python
avg_cost = df.groupby('Medical Condition')['Billing Amount'].mean()
```

```python
sns.barplot(avg_cost)
plt.title('Average cost by disease')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# Histograms

# Age distribution : 

```python
sns.histplot(df['Age'],bins=10,kde=True)
plt.title('Age distribution')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# Medical-cost distribution

```python
sns.histplot(df['Billing Amount'],bins=20,kde=True)
plt.title('Medical-cost distribution')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# Pie charts : 

# Gender distribution

```python
age_distribution = df['Gender'].value_counts().reset_index()
age_distribution.columns = ['Gender','count']
```

```python
fig = px.pie(age_distribution,names='Gender',values='count',title='Gender Distribution',hole=0.5)
fig.show()
```

# Insurance distribution

```python
insurance_dist = df['Insurance Provider'].value_counts().reset_index()
insurance_dist.columns = ['insurance','count']
```

```python
fig = px.pie(insurance_dist,names='insurance',values='count',title='Insurance distribution',hole=0.5)
fig.show()
```

# Box plots : 

# Medical cost by gender

```python
sns.boxplot(data=df,x='Gender',y = 'Billing Amount')
plt.title('Medical cost by gender')
plt.grid(axis='y',linestyle = '--')
plt.show()
```

# Medical cost by disease

```python

fig = px.box(
    df,
    x="Medical Condition",
    y="Billing Amount",
    title="Medical Cost by Disease"
)
fig.show()
```

# Line charts :

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Date of Admission'].value_counts().sort_index().plot(kind='line', color='teal')
plt.title('Daily Patient Admissions Over Time')
plt.xlabel('Date of Admission')
plt.ylabel('Number of Patients')
plt.show()
```
